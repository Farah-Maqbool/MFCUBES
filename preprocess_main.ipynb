```python
import pandas as pd
import matplotlib.pyplot as plt
import csv
import seaborn as sns
from collections import Counter
from tensorflow.keras.preprocessing.text import Tokenizer
```

API Calls Count

```python
sequence = []
with open("all_analysis_data.txt", "r") as f:
    for line in f:
        seq = line.strip().lower().split()
        sequence.append(seq)

all_apis = [api for seq in sequence for api in seq]

api_counter = Counter(all_apis)

print(f"Total unique APIs: {len(api_counter)}")
api_counter
```

```python
with open("all_analysis_data.csv", "w", newline="") as f:
    writer = csv.writer(f, delimiter=" ")

    for seq in sequence:
        writer.writerow(seq)
```

Load data

```python
labels = pd.read_csv("labels.csv", header=None)
df = pd.read_csv("all_analysis_data.csv", header=None)
df.columns = ["API_Calls"]
df["API_Labels"] = labels
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
def collapse_duplicates(seq):
    if not seq:
        return []
    collapsed = [seq[0]]
    for token in seq[1:]:
        if token != collapsed[-1]:
            collapsed.append(token)
    return collapsed

# Apply to all sequences
sequences = [collapse_duplicates(seq) for seq in df.API_Calls]

print(f"Example after collapsing duplicates: {sequences[0][:20]}")
```
